```python
# Import foundational libraries
import pandas as pd
import numpy as np

# Load the anonymized financial dataset
df = pd.read_csv('/content/creditcard.csv')

# Prove the class imbalance (0 = Normal, 1 = Fraud)
print("Class Imbalance:")
print(df['Class'].value_counts())
```

```python
# Create X by dropping the answers so the model can hunt blindly
X = df.drop(columns=['Class'])

# Save the answers separately
y = df['Class']
```

```python
# Import the model blueprint
from sklearn.ensemble import IsolationForest

# Build the model object
model = IsolationForest()

# Train the model and get its predictions (1 for normal, -1 for anomaly)
predictions = model.fit_predict(X)

# Glue the model's predictions back onto our original table
df['Anomaly'] = predictions
```

```python
# Filter the dataframe to only show the anomalies (-1)
fraud_alerts = df[df['Anomaly'] == -1]

print(f"Total suspicious transactions flagged: {len(fraud_alerts)}")

# Display the first 5 flagged transactions for investigation
fraud_alerts.head()
```
